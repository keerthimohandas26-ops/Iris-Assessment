```python
import zipfile
import os

# Unzip the archive if it exists and 'bmi.csv' is not directly available
zip_file_path = '/content/archive.zip'
if os.path.exists(zip_file_path):
    with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
        print("Contents of archive.zip:")
        for file_name in zip_ref.namelist():
            print(file_name)
        # If bmi.csv is not directly in /content/, extract all
        if not os.path.exists('/content/bmi.csv'):
            zip_ref.extractall('/content/')
            print(f"Extracted contents of {zip_file_path} to /content/")
        else:
            print("'bmi.csv' already exists in '/content/'")
else:
    print(f"'{zip_file_path}' not found.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('/content/insurance.csv')
df.head()
df.info()
df.isnull().sum()
df.describe()
df['bmi'].hist() # Assuming 'bmi' column in insurance.csv is lowercase
plt.title('BMI Distribution')
plt.xlabel('BMI')
plt.ylabel('Count')
plt.show()
```
